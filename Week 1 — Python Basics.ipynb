## 3. Input and output

`input()` always returns a string, so convert it with `int()` or `float()`.
The cell below uses a *safe* default so the notebook can also run without typing anything.

```python
def ask(prompt, default, cast=float):
    """Read a value from the user; fall back to a default if input is empty/unavailable."""
    try:
        raw = input(f"{prompt} [default {default}]: ").strip()
    except Exception:
        raw = ""
    return cast(raw) if raw else cast(default)

celsius = ask("Enter temperature in Celsius", 32)
print(f"You entered {celsius} deg C")
```

## 4. Conditional statements (if / elif / else)

```python
def classify_temperature(t):
    if t < 10:
        return "Cold"
    elif t < 25:
        return "Pleasant"
    elif t < 35:
        return "Warm"
    else:
        return "Hot"

for t in [5, 18, 30, 41]:
    print(t, "deg C ->", classify_temperature(t))
```

## 5. Loops (for, while) with break / continue

```python
# for loop
total = 0
for t in [28.4, 31.0, 26.5, 33.2]:
    total += t
print("Sum of readings:", round(total, 2))

# while loop
countdown = 3
while countdown > 0:
    print("Starting in", countdown)
    countdown -= 1

# break and continue
for n in range(1, 11):
    if n % 2 == 0:
        continue          # skip even numbers
    if n > 7:
        break             # stop early
    print("odd:", n)
```

## Hands-On 1 — Temperature converter

```python
def c_to_f(c): return c * 9 / 5 + 32
def f_to_c(f): return (f - 32) * 5 / 9
def c_to_k(c): return c + 273.15

def temperature_converter(value, unit):
    unit = unit.strip().upper()
    if unit == "C":
        return {"Fahrenheit": round(c_to_f(value), 2), "Kelvin": round(c_to_k(value), 2)}
    elif unit == "F":
        c = f_to_c(value)
        return {"Celsius": round(c, 2), "Kelvin": round(c_to_k(c), 2)}
    else:
        return {"error": "Unit must be C or F"}

print(temperature_converter(37, "C"))
print(temperature_converter(98.6, "F"))
print(temperature_converter(10, "X"))
```

## Hands-On 2 — Calculator

```python
def calculator(x, y, op):
    if op == "+": return x + y
    if op == "-": return x - y
    if op == "*": return x * y
    if op == "/":
        return "Error: division by zero" if y == 0 else x / y
    if op == "%":
        return "Error: modulo by zero" if y == 0 else x % y
    if op == "**": return x ** y
    return "Error: unsupported operator"

for op in ["+", "-", "*", "/", "%", "**", "$"]:
    print(f"12 {op} 5 =", calculator(12, 5, op))
print("9 / 0 =", calculator(9, 0, "/"))
```

## Client Project — Average temperature data processing script

Reads the daily readings CSV using **plain Python only** (no pandas yet), skips bad or
missing values, and reports overall plus city-wise statistics.

```python
import csv
DATA = "data/weather_energy.csv"   # keep the data folder next to this notebook

rows, skipped = [], 0
with open(DATA, newline="") as f:
    for row in csv.DictReader(f):
        try:
            rows.append({"city": row["city"],
                         "date": row["date"],
                         "temp": float(row["temperature_c"])})
        except (ValueError, TypeError):
            skipped += 1        # empty or non-numeric temperature

print("Valid readings:", len(rows), "| Skipped (missing/invalid):", skipped)
```

```python
temps = [r["temp"] for r in rows]
overall_avg = sum(temps) / len(temps)

city_totals, city_counts = {}, {}
for r in rows:
    city_totals[r["city"]] = city_totals.get(r["city"], 0) + r["temp"]
    city_counts[r["city"]] = city_counts.get(r["city"], 0) + 1

print(f"Overall average temperature : {overall_avg:.2f} deg C")
print(f"Minimum / Maximum           : {min(temps):.1f} / {max(temps):.1f} deg C")
print("\nCity-wise average temperature")
print("-" * 40)
for city in sorted(city_totals):
    avg = city_totals[city] / city_counts[city]
    bar = "#" * int(avg / 2)
    print(f"{city:<11} {avg:6.2f} deg C  n={city_counts[city]:<4} {bar}")
```

```python
# Warmest and coldest recorded day
hottest = max(rows, key=lambda r: r["temp"])
coldest = min(rows, key=lambda r: r["temp"])
print(f"Hottest day: {hottest['date']} in {hottest['city']} at {hottest['temp']} deg C")
print(f"Coldest day: {coldest['date']} in {coldest['city']} at {coldest['temp']} deg C")

with open("week1_temperature_report.txt", "w") as f:
    f.write("AVERAGE TEMPERATURE REPORT\n")
    f.write(f"Valid readings: {len(rows)} (skipped {skipped})\n")
    f.write(f"Overall average: {overall_avg:.2f} C\n")
    for city in sorted(city_totals):
        f.write(f"{city}: {city_totals[city] / city_counts[city]:.2f} C\n")
print("\nSaved -> week1_temperature_report.txt")
```

## Summary of concepts learned — Week 1

- **Variables and data types**: Python is dynamically typed; `int`, `float`, `str`, `bool`,
  `list` and `dict` cover most day-to-day data work, and `type()` confirms what an object is.
- **Operators**: arithmetic (`//` floor division, `%` remainder, `**` power), comparison,
  logical (`and`, `or`, `not`) and compound assignment (`+=`, `*=`).
- **Input / output**: `input()` returns a string, so numeric input must be cast; `print()`
  with f-strings gives clean, formatted reports.
- **Conditionals**: `if / elif / else` chains map continuous values onto labelled categories,
  which is the basis of rule-based data classification.
- **Loops**: `for` iterates over sequences, `while` repeats until a condition fails, and
  `break` / `continue` control the flow early.
- **Functions and error handling**: small single-purpose functions (`c_to_f`, `calculator`)
  keep logic reusable, and `try / except` lets a data script survive bad rows instead of crashing.
- **Client project takeaway**: a real data pipeline is read -> validate -> aggregate -> report.
  Missing values existed in the raw file and had to be skipped explicitly before averaging.
