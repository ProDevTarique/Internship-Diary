## 3. Vectorised operations and aggregation

```python
x = np.array([28.4, 31.0, 26.5, 33.2, 22.1])
print("x + 5      :", x + 5)
print("to Fahrenheit:", np.round(x * 9 / 5 + 32, 1))
print("sqrt       :", np.round(np.sqrt(x), 2))
print("mean/median/std:", round(x.mean(), 2), np.median(x), round(x.std(ddof=1), 2))

m = np.arange(1, 13).reshape(3, 4)
print("column sums:", m.sum(axis=0))
print("row means  :", m.mean(axis=1))
print("cumsum     :", x.cumsum().round(1))
print("argmax     :", x.argmax(), "->", x[x.argmax()])
```

## 4. Broadcasting

```python
matrix = np.array([[1., 2., 3.], [4., 5., 6.]])
row = np.array([10., 20., 30.])       # shape (3,)  -> stretched across rows
col = np.array([[100.], [200.]])      # shape (2,1) -> stretched across columns

print(matrix + row)
print(matrix + col)

# practical use: standardise each column (z-score)
data = np.array([[28., 60.], [31., 70.], [26., 55.], [33., 80.]])
print("\nz-scores:\n", np.round((data - data.mean(0)) / data.std(0), 2))
```

## Client Project — clean and aggregate with Pandas

Same pipeline as Week 2 but vectorised: drop duplicates, handle missing values,
add derived columns, then group and pivot.

```python
# --- 1. remove duplicates ---
clean = df.drop_duplicates(subset=[c for c in df.columns if c != "record_id"]).copy()
print("Duplicates removed:", len(df) - len(clean), "-> rows:", len(clean))

# --- 2. handle missing values ---
num_cols = ["temperature_c", "humidity", "wind_kmph", "energy_units"]
print("\nMissing before:\n", clean[num_cols].isna().sum())

# strategy: temperature is the target -> drop those rows;
#           other features -> fill with the city median (group-aware imputation)
clean = clean.dropna(subset=["temperature_c"])
for col in ["humidity", "wind_kmph", "energy_units"]:
    clean[col] = clean.groupby("city")[col].transform(lambda s: s.fillna(s.median()))
print("\nMissing after:\n", clean[num_cols].isna().sum())
print("Rows now:", len(clean))
```

```python
# --- 3. derived columns ---
clean["temp_f"] = (clean.temperature_c * 9 / 5 + 32).round(1)
clean["month"] = clean.date.dt.month_name().str[:3]
clean["comfort"] = pd.cut(clean.temperature_c,
                          bins=[-np.inf, 15, 25, 33, np.inf],
                          labels=["Cold", "Pleasant", "Warm", "Hot"])
clean["energy_per_degree"] = (clean.energy_units / clean.temperature_c.abs()).round(2)
clean[["city", "date", "temperature_c", "temp_f", "month", "comfort"]].head()
```

```python
# --- 4. group-by aggregation ---
summary = (clean.groupby("city")
           .agg(readings=("temperature_c", "size"),
                avg_temp=("temperature_c", "mean"),
                max_temp=("temperature_c", "max"),
                avg_humidity=("humidity", "mean"),
                avg_energy=("energy_units", "mean"))
           .round(2)
           .sort_values("avg_temp", ascending=False))
summary
```

```python
# multi-level grouping and pivot table
monthly = clean.groupby(["city", "month"])["temperature_c"].mean().round(2)
print("Multi-index result (head):\n", monthly.head(8))

order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul"]
pivot = clean.pivot_table(index="city", columns="month",
                          values="temperature_c", aggfunc="mean").round(1)
pivot = pivot[[m for m in order if m in pivot.columns]]
pivot
```

```python
# comfort distribution per city (%)
dist = pd.crosstab(clean.city, clean.comfort, normalize="index").mul(100).round(1)
print(dist)

# --- 5. merge with an external reference table ---
regions = pd.DataFrame({"city": ["Patna", "Delhi", "Mumbai", "Bengaluru", "Kolkata"],
                        "region": ["East", "North", "West", "South", "East"]})
merged = clean.merge(regions, on="city", how="left")
print("\nRegion-wise average temperature:\n",
      merged.groupby("region")["temperature_c"].mean().round(2))
```

```python
# --- 6. save outputs for Week 4 ---
clean.to_csv("week3_clean_dataset.csv", index=False)
summary.to_csv("week3_city_summary.csv")
print("Saved -> week3_clean_dataset.csv and week3_city_summary.csv")
print("Final shape:", clean.shape)
```

## Summary of concepts learned — Week 3

**NumPy**
- `np.array`, `arange`, `linspace`, `zeros/ones/eye` and the `default_rng` generator create
  arrays; `shape`, `dtype`, `ndim` and `size` describe them.
- Vectorised arithmetic replaces Python loops and is far faster because it runs in compiled C.
- Boolean masks (`m[m > 6]`) and fancy indexing select data without loops.
- Aggregations accept `axis=0` (down columns) or `axis=1` (across rows); `nanmean` ignores NaN.
- **Broadcasting** lets arrays of different shapes combine by stretching size-1 dimensions —
  the mechanism behind one-line z-score standardisation.

**Pandas**
- A `Series` is a labelled 1-D array; a `DataFrame` is a table of aligned Series.
- `read_csv(parse_dates=...)`, `head`, `info`, `describe`, `isna().sum()` and `duplicated()`
  form the standard first-look routine.
- `loc` selects by label, `iloc` by position; boolean masks, `query()` and `isin()` filter rows.
- Missing data: `dropna` for the target column, and `groupby(...).transform(fillna(median))`
  for group-aware imputation, which is more accurate than a single global median.
- `pd.cut` bins continuous values into categories; `dt` accessors extract date parts.
- `groupby().agg()` with named aggregations, `pivot_table`, `crosstab` and `merge` cover
  almost all reshaping and joining needs.
- **Client project takeaway**: the whole Week 2 cleaning script collapsed into a handful of
  vectorised Pandas calls, with clearer intent and far fewer lines.
