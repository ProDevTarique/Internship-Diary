## 4. Sets (unique values and set algebra)

## 5. Functions: positional, default, *args, **kwargs

```python
def summarise(values, ndigits=2, label="dataset"):
    """Return count, mean, min and max of a numeric list."""
    return {"label": label,
            "count": len(values),
            "mean": round(sum(values) / len(values), ndigits),
            "min": min(values), "max": max(values)}

print(summarise([28.4, 31.0, 26.5, 33.2], label="temps"))

def total(*args, **kwargs):
    print("args  :", args)
    print("kwargs:", kwargs)
    return sum(args)

print("Total :", total(1, 2, 3, unit="celsius"))
```

## 6. Lambda with map / filter / sorted / reduce

```python
from functools import reduce

temps = [28.4, 31.0, 26.5, 33.2, 22.1, 35.8]
to_f = lambda c: round(c * 9 / 5 + 32, 1)

print("Mapped to F :", list(map(to_f, temps)))
print("Filtered >30:", list(filter(lambda t: t > 30, temps)))
print("Product     :", reduce(lambda x, y: x * y, [1, 2, 3, 4]))

people = [("Aman", 24), ("Riya", 21), ("Zaid", 29)]
print("Sorted by age:", sorted(people, key=lambda p: p[1]))
```

## 7. Recursion

```python
def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)

def fib(n, memo={}):
    if n < 2: return n
    if n not in memo:
        memo[n] = fib(n - 1, memo) + fib(n - 2, memo)
    return memo[n]

def flatten(nested):
    out = []
    for item in nested:
        out.extend(flatten(item)) if isinstance(item, list) else out.append(item)
    return out

print("factorial(6) :", factorial(6))
print("fib series   :", [fib(i) for i in range(10)])
print("flatten      :", flatten([1, [2, [3, [4, 5]], 6], 7]))
```

## 8. Comprehensions (list, dict, set, nested)

```python
nums = list(range(1, 11))
print("Squares          :", [n ** 2 for n in nums])
print("Even squares     :", [n ** 2 for n in nums if n % 2 == 0])
print("Sum of squares   :", sum(n ** 2 for n in nums))
print("Dict comprehension:", {n: n ** 3 for n in range(1, 6)})
print("Set comprehension :", {n % 3 for n in nums})
print("Nested (pairs)    :", [(a, b) for a in "AB" for b in [1, 2]])
```

## Hands-On — data transformation helpers

```python
def sum_of_squares(values):
    return sum(v ** 2 for v in values)

def filter_range(values, low, high):
    return [v for v in values if low <= v <= high]

def normalise(values):
    lo, hi = min(values), max(values)
    return [round((v - lo) / (hi - lo), 3) for v in values]

def group_by(records, key):
    out = {}
    for r in records:
        out.setdefault(r[key], []).append(r)
    return out

temps = [28.4, 31.0, 26.5, 33.2, 22.1, 35.8]
print("Sum of squares 1..10 :", sum_of_squares(range(1, 11)))
print("In range 26-34       :", filter_range(temps, 26, 34))
print("Normalised           :", normalise(temps))
```

## Client Project — Data cleaning script

Steps: load CSV -> drop exact duplicate records -> drop rows with missing values ->
cast types -> filter to a valid range -> remove statistical outliers -> save a clean file.

```python
import csv
DATA = "data/weather_energy.csv"   # keep the data folder next to this notebook
NUMERIC = ["temperature_c", "humidity", "wind_kmph", "energy_units"]

with open(DATA, newline="") as f:
    raw = list(csv.DictReader(f))
print("Rows loaded:", len(raw))

# --- 1. remove duplicates (ignore record_id, compare the actual content) ---
seen, deduped = set(), []
for row in raw:
    fingerprint = tuple(row[c] for c in row if c != "record_id")
    if fingerprint not in seen:
        seen.add(fingerprint)
        deduped.append(row)
print("After removing duplicates:", len(deduped), f"(removed {len(raw) - len(deduped)})")
```

```python
# --- 2. drop rows with missing values and cast numeric columns ---
clean = []
for row in deduped:
    if any(row[c] in ("", None) for c in NUMERIC):
        continue
    rec = {"record_id": int(row["record_id"]), "city": row["city"], "date": row["date"]}
    rec.update({c: float(row[c]) for c in NUMERIC})
    clean.append(rec)
print("After dropping missing values:", len(clean))

# --- 3. range filter (plausible physical values) ---
clean = [r for r in clean if -10 <= r["temperature_c"] <= 55 and 0 <= r["humidity"] <= 100]
print("After range filter:", len(clean))
```

```python
# --- 4. outlier removal using the 1.5 * IQR rule ---
def quantile(sorted_vals, q):
    pos = (len(sorted_vals) - 1) * q
    lo, hi = int(pos), min(int(pos) + 1, len(sorted_vals) - 1)
    return sorted_vals[lo] + (sorted_vals[hi] - sorted_vals[lo]) * (pos - lo)

vals = sorted(r["temperature_c"] for r in clean)
q1, q3 = quantile(vals, 0.25), quantile(vals, 0.75)
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
final = [r for r in clean if low <= r["temperature_c"] <= high]
print(f"Q1={q1:.2f} Q3={q3:.2f} IQR={iqr:.2f} -> keep [{low:.2f}, {high:.2f}]")
print("Final clean rows:", len(final), f"(outliers removed: {len(clean) - len(final)})")
```

```python
# --- 5. city-wise summary and save ---
groups = group_by(final, "city")
print(f"{'City':<11}{'Count':>6}{'Avg temp':>10}{'Avg energy':>12}")
print("-" * 39)
for city in sorted(groups):
    g = groups[city]
    print(f"{city:<11}{len(g):>6}{sum(r['temperature_c'] for r in g) / len(g):>10.2f}"
          f"{sum(r['energy_units'] for r in g) / len(g):>12.2f}")

with open("week2_cleaned_data.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(final[0].keys()))
    writer.writeheader(); writer.writerows(final)
print("\nSaved -> week2_cleaned_data.csv")
```

## Summary of concepts learned — Week 2

- **Lists** are ordered and mutable: `append`, `insert`, `remove`, slicing and `sorted()`
  cover most sequence work.
- **Tuples** are immutable and therefore safe as fixed records and dictionary keys; unpacking
  (`city, date, temp = reading`) makes code readable.
- **Dictionaries** give O(1) key lookup; `get()` with a default avoids `KeyError`, and
  `setdefault()` is the idiomatic way to build grouped data.
- **Sets** remove duplicates instantly and support union / intersection / difference, which is
  why the cleaning script used a `set` of row fingerprints for de-duplication.
- **Functions** support defaults, `*args` and `**kwargs`; docstrings document intent.
- **Lambda** creates short throwaway functions, most useful as the `key=` of `sorted()` or
  inside `map` / `filter` / `reduce`.
- **Recursion** solves self-similar problems (factorial, Fibonacci, flattening nested lists);
  memoisation removes the exponential cost.
- **Comprehensions** replace loop-and-append with one expressive line, and generator
  expressions inside `sum()` avoid building an intermediate list.
- **Client project takeaway**: cleaning order matters — de-duplicate, then handle missing
  values, then filter invalid ranges, and only then remove outliers with the 1.5*IQR rule.
